
# Documentación del mini–proyecto: Consumo de APIs con TypeScript y CodeSandbox

Este cuaderno documenta, la aplicación web que se construyó en CodeSandbox para demostrar:

1. Cómo consumir una **API pública** (Quotable) desde el navegador.
2. Cómo consumir una **API externa con API key** (OpenWeatherMap) desde el navegador.
3. Cómo estructurar el código en una plantilla de **TypeScript Vanilla** con estos archivos:

- `index.html` – estructura del documento HTML.
- `src/styles.css` – estilos visuales (colores, espaciados, tipografía).
- `src/index.ts` – lógica de la aplicación escrita en TypeScript.

En esta aplicación, el usuario puede:

- Pulsar un botón para obtener una **frase aleatoria** desde `https://api.quotable.io/random`.
- Introducir el nombre de una ciudad y obtener el **clima actual**, usando la API de OpenWeatherMap.

A continuación se describe cada archivo, su función y luego se muestra el código totalmente comentado.



## 1. Archivo `index.html` – Estructura del documento

El archivo `index.html` define la estructura base de la página web. No contiene estilos ni lógica de negocio; su objetivo principal es:

- Declarar el idioma y la codificación del documento.
- Definir la estructura de las secciones visibles: título principal, sección de frases y sección de clima.
- Incluir el script principal escrito en TypeScript (`src/index.ts`) como módulo ES (`type="module"`).

A continuación se muestra el contenido del archivo con comentarios en línea.


````
<!-- index.html -->

<!DOCTYPE html>
<html lang="es"> <!-- Se indica que el contenido está en español -->
  <head>
    <meta charset="UTF-8" /> <!-- Codificación UTF-8 para soportar acentos y caracteres especiales -->
    <title>Demo APIs con TypeScript</title> <!-- Título que se verá en la pestaña del navegador -->
  </head>
  <body>
    <!-- Título principal de la página -->
    <h1>Demo APIs</h1>

    <!-- ===============================
         Sección 1: Frase aleatoria
         =============================== -->
    <section>
      <!-- Subtítulo explicando que esta parte consume una API pública -->
      <h2>1. Frase al azar desde API pública</h2>

      <!-- Botón que el usuario pulsará para obtener una nueva frase -->
      <button id="btn-quote">Obtener frase</button>

      <!-- Contenedor donde se imprimirá el contenido de la frase -->
      <p id="quote"></p>

      <!-- Contenedor donde se imprimirá el autor de la frase -->
      <p id="author"></p>

      <!-- Contenedor para mostrar mensajes de error relacionados con la frase -->
      <p id="error"></p>
    </section>

    <!-- ===============================
         Sección 2: Clima con OpenWeather
         =============================== -->
    <section>
      <!-- Subtítulo explicando qué hace esta sección -->
      <h2>2. Clima (OpenWeatherMap)</h2>

      <!-- Campo de texto donde el usuario escribe la ciudad -->
      <input id="city-input" placeholder="Ej: Bogotá" />

      <!-- Botón que dispara la consulta del clima -->
      <button id="btn-weather">Obtener clima</button>

      <!-- Contenedor donde se mostrará la información de clima formateada -->
      <pre id="weather-result"></pre>

      <!-- Contenedor donde se mostrarán mensajes de error específicos del clima -->
      <pre id="weather-error"></pre>
    </section>

    <!-- Inclusión del script principal escrito en TypeScript.
         Se usa type="module" para poder utilizar import/export en el código TS.
         Nótese que el archivo está dentro de la carpeta src, por eso la ruta ./src/index.ts -->
    <script type="module" src="./src/index.ts"></script>
  </body>
</html>
````


## 2. Archivo `src/styles.css` – Estilos y colores

El archivo `styles.css` se encarga de la apariencia visual de la aplicación. Define:

- Tipografía y colores globales.
- El estilo de cada sección (tarjetas, sombras, bordes redondeados).
- El aspecto de los botones, campos de texto y mensajes de error.
- El formato visual del texto de la frase y del autor.

En la plantilla de TypeScript usada en CodeSandbox, el CSS no se enlaza con `<link>` dentro del HTML, sino que se **importa desde el archivo TypeScript** (`import "./styles.css";`), permitiendo que el bundler inserte los estilos de manera automática.

A continuación se muestra el contenido de `src/styles.css`, comentado sección por sección.


````
/* src/styles.css */

/* ===========================
   ESTILOS GENERALES DEL CUERPO
   =========================== */
body {
  /* Fuente general de la página */
  font-family: system-ui, sans-serif;

  /* Centra el contenido y limita el ancho máximo */
  margin: 2rem auto;
  max-width: 700px;

  /* Define el interlineado para mejorar la legibilidad */
  line-height: 1.6;

  /* Color de fondo gris claro para dar contraste con las tarjetas blancas */
  background: #f4f6f9;

  /* Color de texto principal */
  color: #333;
}

/* ===========================
   SECCIONES (TARJETAS)
   =========================== */
section {
  /* Fondo blanco para que parezcan tarjetas sobre el fondo gris */
  background: white;

  /* Espaciado interno */
  padding: 1.5rem;

  /* Bordes redondeados para una apariencia más moderna */
  border-radius: 12px;

  /* Borde suave alrededor de la tarjeta */
  border: 1px solid #e1e4ea;

  /* Separación entre tarjetas */
  margin-bottom: 1.5rem;

  /* Sombra ligera para dar sensación de profundidad */
  box-shadow: 0 2px 8px rgba(0, 0, 0, 0.05);
}

/* Colores de los títulos principales y secundarios */
h1,
h2 {
  color: #1e2a55;
}

/* ===========================
   BOTONES
   =========================== */
button {
  /* Tamaño y espaciado del botón */
  padding: 0.6rem 1.2rem;

  /* Manito al pasar sobre el botón para indicar que es interactivo */
  cursor: pointer;

  /* Color de fondo principal del botón */
  background: #1e2a55;

  /* Sin borde nativo del navegador */
  border: none;

  /* Bordes redondeados */
  border-radius: 6px;

  /* Color del texto del botón */
  color: white;

  /* Transición suave al hacer hover */
  transition: 0.2s ease-in-out;
}

/* Efecto al pasar el ratón por encima del botón */
button:hover {
  background: #2d3c7c;
}

/* ===========================
   CAMPOS DE TEXTO (INPUT)
   =========================== */
input {
  /* Espaciado interno para que el texto no quede pegado al borde */
  padding: 0.5rem;

  /* Borde suave alrededor del input */
  border: 1px solid #ccd2dd;

  /* Bordes redondeados para mantener el estilo visual */
  border-radius: 6px;

  /* Pequeña separación con respecto al botón */
  margin-right: 0.4rem;
}

/* ===========================
   MENSAJES DE ERROR
   =========================== */
#error,
#weather-error {
  /* Color rojo para indicar claramente un error */
  color: #c62828;

  /* Negrita para llamar la atención del usuario */
  font-weight: bold;

  /* Pequeño margen superior */
  margin-top: 0.5rem;
}

/* ===========================
   RESULTADOS DE FRASES
   =========================== */

/* Texto de la frase en cursiva para darle énfasis */
#quote {
  font-style: italic;
  margin-top: 1rem;
}

/* Autor alineado a la derecha para simular una cita formal */
#author {
  text-align: right;
  font-weight: bold;
  color: #555;
}
````


## 3. Archivo `src/index.ts` – Lógica de la aplicación

El archivo `index.ts` contiene toda la lógica de la aplicación escrita en TypeScript. Sus responsabilidades principales son:

1. **Importar los estilos CSS** para que formen parte del bundle.
2. **Referenciar los elementos del DOM** (botones, campos de texto, contenedores de resultados).
3. Implementar la función `loadRandomQuote()` para:
   - Llamar a la API pública `https://api.quotable.io/random`.
   - Manejar errores de red o de servidor.
   - Mostrar la frase y su autor en la interfaz.
4. Implementar la función `loadWeather()` para:
   - Leer la ciudad introducida por la persona usuaria.
   - Construir la URL de la API de OpenWeatherMap.
   - Llamar a la API con `fetch`, utilizando la API key personal.
   - Manejar errores y mostrar el resultado formateado.
5. Registrar los listeners de eventos de clic en los botones.

A continuación se muestra el código completo, muy comentado línea por línea.


````
// src/index.ts

// ============================================================================
// Importación de estilos
// ============================================================================

/**
 * Se importa el archivo de estilos CSS que se encuentra en el mismo directorio
 * que este archivo TypeScript (src/styles.css).
 *
 * En la plantilla de CodeSandbox, el bundler se encarga de procesar este
 * import, generar el CSS final y adjuntarlo al documento HTML.
 */
import "./styles.css";

// ============================================================================
// Referencias a elementos del DOM
// ============================================================================

/**
 * A continuación se obtienen las referencias a los elementos HTML mediante
 * `document.querySelector`. Se especifican tipos genéricos de TypeScript
 * (por ejemplo, HTMLButtonElement) para que el compilador sepa qué propiedades
 * existen en cada referencia y ofrezca autocompletado.
 *
 * El operador `!` (non-null assertion) indica al compilador que se asume
 * que el elemento existe en el DOM. Esto es correcto en este caso porque
 * el HTML está bajo nuestro control y se sabe que los IDs están presentes.
 */

// Botón que dispara la carga de una frase aleatoria
const btnQuote = document.querySelector<HTMLButtonElement>("#btn-quote")!;

// Contenedor donde se mostrará el texto de la frase
const quoteEl = document.querySelector<HTMLParagraphElement>("#quote")!;

// Contenedor donde se mostrará el autor de la frase
const authorEl = document.querySelector<HTMLParagraphElement>("#author")!;

// Contenedor para mostrar errores relacionados con la obtención de frases
const errorEl = document.querySelector<HTMLParagraphElement>("#error")!;

// Botón que dispara la consulta del clima
const btnWeather =
  document.querySelector<HTMLButtonElement>("#btn-weather")!;

// Campo de texto donde se escribe el nombre de la ciudad
const cityInput = document.querySelector<HTMLInputElement>("#city-input")!;

// Contenedor donde se mostrará el resultado del clima
const weatherResult = document.querySelector<HTMLElement>("#weather-result")!;

// Contenedor donde se imprimirán errores relacionados con el clima
const weatherError = document.querySelector<HTMLElement>("#weather-error")!;

// ============================================================================
// Tipos auxiliares para las respuestas de las APIs
// ============================================================================

/**
 * Tipo que modela la respuesta simplificada de la API de Quotable.
 * La API devuelve más campos, pero en esta aplicación solo se utilizan
 * `content` (texto de la frase) y `author`.
 */
type QuotableResponse = {
  content: string;
  author: string;
};

/**
 * Tipo que modela la respuesta simplificada de la API de OpenWeatherMap.
 * De nuevo, la respuesta real contiene muchos más campos, pero aquí solo
 * se utilizan algunos para mostrar la información más relevante.
 */
type WeatherResponse = {
  name: string;
  weather: { description: string }[];
  main: { temp: number };
};

// ============================================================================
// Función auxiliar genérica para llamar APIs que devuelven JSON
// ============================================================================

/**
 * Esta función genérica recibe una URL y devuelve una promesa del tipo `T`,
 * donde `T` representa la forma del objeto JSON esperado. De esta forma,
 * se puede reutilizar el mismo patrón de llamada para diferentes APIs,
 * manteniendo tipado el resultado (por ejemplo: QuotableResponse, WeatherResponse).
 */
async function callJsonApi<T>(url: string): Promise<T> {
  // Se realiza la solicitud HTTP usando `fetch`
  const res = await fetch(url);

  // Si el código de estado no indica éxito (2xx), se lanza un error
  if (!res.ok) {
    throw new Error(`HTTP ${res.status} en ${url}`);
  }

  // Se parsea la respuesta como JSON y se castea al tipo genérico T
  return (await res.json()) as T;
}

// ============================================================================
// 1. Función loadRandomQuote: consumir la API pública de frases
// ============================================================================

/**
 * Esta función se ejecuta cuando la persona usuaria hace clic en el botón
 * "Obtener frase". Su objetivo es llamar a la API de Quotable y actualizar
 * la interfaz con el resultado.
 */
async function loadRandomQuote() {
  // Se resetean los mensajes previos y se muestra un indicador de carga
  quoteEl.textContent = "Cargando...";
  authorEl.textContent = "";
  errorEl.textContent = "";

  try {
    // Se llama a la API usando la función genérica callJsonApi, especificando
    // que se espera un resultado del tipo QuotableResponse.
    const data = await callJsonApi<QuotableResponse>(
      "https://api.quotable.io/random"
    );

    // Si la respuesta es correcta, se actualizan los elementos del DOM
    quoteEl.textContent = `"${data.content}"`;
    authorEl.textContent = `— ${data.author}`;
  } catch (err) {
    // Si ocurre cualquier error (problema de red, servidor caído, etc.),
    // se muestra un mensaje de error al usuario y se loguea en consola.
    console.error(err);
    quoteEl.textContent = "";
    authorEl.textContent = "";
    errorEl.textContent =
      "No se pudo obtener la frase. ¿Hay conexión a internet?";
  }
}

// ============================================================================
// 2. Función loadWeather: consumir la API de OpenWeatherMap
// ============================================================================

/**
 * API key personal de OpenWeatherMap. En un entorno de producción,
 * esta clave no debería exponerse en el código cliente, pero para
 * fines educativos y demostrativos es aceptable.
 */
const OPENWEATHER_API_KEY = "3c8c0a24be0fbc288e24c9e82fee7c26";

/**
 * Esta función se ejecuta cuando la persona usuaria hace clic en el botón
 * "Obtener clima". Su objetivo es:
 *
 * 1. Leer la ciudad introducida en el campo de texto.
 * 2. Construir la URL para la API de OpenWeatherMap con dicha ciudad,
 *    incluyendo la API key y los parámetros de unidades e idioma.
 * 3. Llamar a la API usando la función genérica callJsonApi<WeatherResponse>.
 * 4. Mostrar el resultado formateado o, en su defecto, un mensaje de error.
 */
async function loadWeather() {
  // Se obtiene la ciudad eliminando espacios al inicio y al final
  const city = cityInput.value.trim();

  // Se borran resultados y errores anteriores
  weatherResult.textContent = "";
  weatherError.textContent = "";

  // Validación básica: si la ciudad está vacía, se informa al usuario y se detiene
  if (!city) {
    weatherError.textContent = "Escribe una ciudad";
    return;
  }

  // Construcción de la URL de la API con parámetros:
  // - q: nombre de la ciudad
  // - appid: API key personal
  // - units=metric: unidades en grados Celsius
  // - lang=es: respuestas en español
  const url =
    "https://api.openweathermap.org/data/2.5/weather" +
    `?q=${encodeURIComponent(city)}` +
    `&appid=${OPENWEATHER_API_KEY}` +
    "&units=metric&lang=es";

  try {
    // Se notifica al usuario que la aplicación está consultando el servidor
    weatherResult.textContent = "Cargando...";

    // Se llama a la API y se indica que se espera un WeatherResponse
    const data = await callJsonApi<WeatherResponse>(url);

    // Se extraen los campos relevantes de la respuesta
    const descripcion =
      data.weather?.[0]?.description ?? "Sin descripción disponible";
    const temp = data.main?.temp ?? NaN;

    // Se construye un texto legible para el usuario
    weatherResult.textContent =
      `Ciudad: ${data.name}\n` +
      `Descripción: ${descripcion}\n` +
      `Temperatura: ${isNaN(temp) ? "?" : temp + " °C"}`;
  } catch (err) {
    // Si hay error (ciudad inexistente, API key inválida, fallo de red, etc.)
    // se muestra un mensaje genérico y se registra el error en consola.
    console.error(err);
    weatherResult.textContent = "";
    weatherError.textContent =
      "No se pudo obtener el clima. Revisa ciudad, API key o conexión.";
  }
}

// ============================================================================
// Registro de manejadores de eventos
// ============================================================================

/**
 * Finalmente, se registran los eventos de clic para los dos botones
 * principales de la interfaz. De este modo, cada vez que el usuario
 * hace clic en un botón, se invoca la función correspondiente.
 */

// Al hacer clic en "Obtener frase" se ejecuta loadRandomQuote
btnQuote.addEventListener("click", loadRandomQuote);

// Al hacer clic en "Obtener clima" se ejecuta loadWeather
btnWeather.addEventListener("click", loadWeather);
````


## 4. Resumen conceptual

En resumen, este mini–proyecto enseñado en CodeSandbox ilustra varios conceptos clave del desarrollo web en el navegador:

1. **Consumo de APIs REST** desde el cliente usando `fetch` y `async/await`.
2. Manejo básico de **errores de red y de servidor**, mostrando mensajes amigables al usuario.
3. Uso de **TypeScript en el frontend**, aprovechando:
   - Tipado estático.
   - Tipos genéricos para modelar respuestas de APIs (`QuotableResponse`, `WeatherResponse`).
   - Anotaciones de tipo para elementos del DOM.
4. Separación de responsabilidades en tres archivos:
   - `index.html` para estructura.
   - `styles.css` para estilos.
   - `index.ts` para lógica y comportamiento.

La misma estructura puede extenderse fácilmente para:

- Añadir más APIs (por ejemplo, de tipo de cambio).
- Incorporar una API propia alojada en un servidor Node/Express.
- Manejar estados más complejos en la interfaz de usuario.

**TypeScript**

**1. Un “superset” de JavaScript**

TypeScript no reemplaza JavaScript; lo amplía.

Todo el código JavaScript válido también es válido en TypeScript.
La diferencia es que TypeScript agrega características adicionales que ayudan a detectar errores antes de ejecutar el programa.

**2. Tipado estático**

La característica más importante.

En JavaScript solo sabes si hay un error cuando ejecutas el código.

En TypeScript puedes declarar tipos:


```
let nombre: string = "Christian";
let edad: number = 29;
let activo: boolean = true;
```


Si se intenta hacer algo incorrecto:
```
edad = "hola"; // Error de TypeScript
```

El error aparece antes de ejecutar, lo que evita fallos en producción.

**Fetch**

Fetch es una función nativa de JavaScript que permite hacer peticiones HTTP (GET, POST, PUT, DELETE, etc.) para comunicarse con APIs o servidores. Es la forma moderna y estándar de traer datos desde internet o enviar información.

Es decir:

**fetch =** herramienta para pedir datos a una URL y recibir una respuesta.
